```python
import time
```

```python
# Bubble Sort
# Time complexity - O(n^2), space - O(1) 

A = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def bubbleSort(arr):
    n = len(arr)
    flag = True
    while flag:
        flag = False
        for i in range(1, n):
            if arr[i - 1] > arr[i]:
                flag = True
                arr[i - 1], arr[i] = arr[i], arr[i - 1];


start_time = time.perf_counter()
sorted_result = bubbleSort(A)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", A)
```

```python
# Insertion Sort
# Time complexity - O(n^2), space - O(1) 

B = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def insertionSort(arr):
    n = len(B)
    for i in range(1,n):
        for j in range(i, 0, -1):
            if arr[j - 1] > arr[j]:
                arr[j - 1], arr[j] = arr[j], arr[j - 1]
            else:
                break


start_time = time.perf_counter()
sorted_result = insertionSort(B)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", B)
```

```python
# Selection Sort
# Time complexity - O(n^2), space - O(1) 

C = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def selectionSort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]


start_time = time.perf_counter()
sorted_result = selectionSort(C)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", C)
```

```python
# Merge Sort - Divide / Conquer
# Time complexity - O(n log n)
# space - O(n) - Can be done in log n but is harder

D = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def mergeSort(arr):
    n = len(arr)
    if n <= 1:
        return arr
    m = n // 2
    L = arr[:m]
    R = arr[m:]
    L = mergeSort(L)
    R = mergeSort(R)
    l, r, i = 0, 0, 0
    L_Len, R_Len = len(L), len(R)
    sortedArr = [0] * n
    while l < L_Len and r < R_Len:
        if L[l] <= R[r]:
            sortedArr[i] = L[l]
            l += 1
        else:
            sortedArr[i] = R[r]
            r += 1
        i += 1
    while l < L_Len:
        sortedArr[i] = L[l]
        i += 1; l += 1
    while r < R_Len:
        sortedArr[i] = R[r]
        i += 1; r += 1
    return sortedArr


start_time = time.perf_counter()
sorted_result = mergeSort(D)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", sorted_result)
```

```python
# Quick Sort
# Time complexity - O(n log n) for the average case, but technically worse case is O(n^2)
# space - O(n) 

E = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def quickSort(arr):
    if len(arr) <= 1:
        return arr
    p = arr[-1]
    L = [x for x in arr[:-1] if x <= p]
    R = [x for x in arr[:-1] if x > p]

    L = quickSort(L)
    R = quickSort(R)

    return L + [p] + R

start_time = time.perf_counter()
sorted_result = quickSort(E)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", sorted_result)
```

```python
# Counting Sort 
# Time complexity - O(n + k) where k is range of data
# space - O(k) 

#Note - This can be written with negative arrays, but we'll stick to +ve arrays. 
# So k is the max of the array

F = [5, 3, 2, 1, 3, 3, 7, 2, 2]

def countingSort(arr):
    n = len(arr)
    maxx = max(arr)
    counts = [0] * (maxx + 1)

    for x in arr:
        counts[x] += 1
    
    i = 0
    for c in range(maxx + 1):
        while counts[c] > 0:
            arr[i] = c
            i += 1
            counts[c] -= 1




start_time = time.perf_counter()
sorted_result = countingSort(F)
end_time = time.perf_counter()

print(f"Time Taken: {end_time - start_time:.6f}")

print("Sorted Array:", F)
```
