### Recursion

#### Structure
- Base Case
    - F(0) = 0...
    - F(1) = 1
- Recursion Case
    - F(n) = F(n-1) + F(n=2)

time complexity - O(2^n) <br>
Space complexity - O(n)

```python
# Fibonacci
# F(0) = 0
# F(1) = 1
# n > 1: F(n) = F(n-1) + F(n-2)

# Time: O(2^n)
def F(n):
    # Base case
    if n == 0:
        return 0
    elif n == 1:
        return 1
    # Recursive Step
    else:
        return F(n-1) + F(n - 2)

F(5)
```

```python
# LinkedList - time: O(n), space: O(n)
class Node:
    def __init__(self, data, next = None):
        self.data = data
        self.next = next

    def __str__(self):
        return str(self.data)
    
def reverse(head):
    if not head:
        return
    reverse(head.next)
    print(head.data)

Head = Node(1)
A = Node(2)
B = Node(3)
C = Node(4)

Head.next = A
A.next = B
B.next = C


reverse(Head)
```
