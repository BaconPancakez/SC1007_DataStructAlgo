# Data Structures: Binary Trees (BT) & Binary Search Trees (BST)

## Introduction

A **Tree** is a fundamental non-linear data structure representing hierarchical data. It consists of **nodes** connected by **edges**. A **Binary Tree (BT)** is a specialized tree where each node has at most two children: a **left child** and a **right child**.

## Binary Tree (BT)

### Definition

A Binary Tree is a finite set of nodes that is either empty or consists of:
1.  A special node called the **Root**.
2.  Two disjoint binary trees called the **left subtree** and the **right subtree**.

Key terms:
* **Parent:** The node directly above a given node.
* **Child:** A node directly below a given node.
* **Leaf:** A node with no children.
* **Internal Node:** A node with at least one child.
* **Height:** The number of edges on the longest path from the root to a leaf.
* **Depth:** The number of edges on the path from the root to a specific node.

### Example BT
        1        <- Root
       / \
      2   3      <- Internal Nodes
     / \   \
    4   5   6    <- Leaves (4, 5, 6)

*Height of this tree is 2.*
*Depth of node 5 is 2.*

### Traversals

Visiting every node in the tree exactly once.

#### 1. Depth-First Search (DFS)

Follows one branch as deep as possible before backtracking. Often implemented recursively (using the call stack) or iteratively using an explicit **Stack**.

* **Pre-Order Traversal** (Root -> Left -> Right)
    1.  Visit the current **Root** node.
    2.  Traverse the **Left** subtree.
    3.  Traverse the **Right** subtree.
    * *Use Case:* Useful for creating a copy of the tree or getting prefix expressions.
    * *Example (for the BT above):* `1, 2, 4, 5, 3, 6`

* **In-Order Traversal** (Left -> Root -> Right)
    1.  Traverse the **Left** subtree.
    2.  Visit the current **Root** node.
    3.  Traverse the **Right** subtree.
    * *Use Case:* Critically important for BSTs as it retrieves elements in sorted order.
    * *Example (for the BT above):* `4, 2, 5, 1, 3, 6`

* **Post-Order Traversal** (Left -> Right -> Root)
    1.  Traverse the **Left** subtree.
    2.  Traverse the **Right** subtree.
    3.  Visit the current **Root** node.
    * *Use Case:* Useful for deleting nodes (process children before the parent) or getting postfix expressions.
    * *Example (for the BT above):* `4, 5, 2, 6, 3, 1`

#### 2. Breadth-First Search (BFS) / Level-Order Traversal

Visits nodes level by level, from top to bottom and left to right within each level. Implemented using a **Queue**.

1.  Add the root node to the queue.
2.  While the queue is not empty:
    * Remove a node from the front of the queue.
    * Visit the node.
    * Add its left child (if any) to the back of the queue.
    * Add its right child (if any) to the back of the queue.
* *Use Case:* Finding the shortest path (in terms of number of edges) from the root to another node. Useful in level-based processing.
* *Example (for the BT above):* `1, 2, 3, 4, 5, 6`

#### Complexity for Traversals (DFS & BFS)

* **Time Complexity:** $O(n)$
    * Each node must be visited exactly once, where $n$ is the number of nodes.

* **Space Complexity:** $O(n)$ (Worst Case)
    * **DFS:** The space is determined by the maximum depth of the recursion stack (or explicit stack). In the worst case (a completely skewed tree, like a linked list), the stack depth can be $O(n)$. For a balanced tree, the height $h$ is $O(\log n)$, so the space complexity is $O(\log n)$.
    * **BFS:** The space is determined by the maximum number of nodes at any level in the queue. In the worst case (a complete or full binary tree), the last level can contain roughly $n/2$ nodes, leading to $O(n)$ space.
    * Therefore, $O(n)$ is the general upper bound for space complexity for both DFS and BFS.

## Binary Search Tree (BST)

### Definition

A Binary Tree where for each node:
1.  The node's value is **greater than** all values in its **left subtree**.
2.  The node's value is **less than** all values in its **right subtree**.
3.  Both the left and right subtrees are also Binary Search Trees.

* **Note on Duplicates:** The strict definition often assumes unique values. If duplicates are allowed, a convention must be chosen (e.g., duplicates go to the right subtree, or nodes store counts/lists). The definition above assumes unique values or that duplicates are consistently placed (e.g., treated as "greater than or equal to" for the right subtree).

### Example BST

          8        <-- Root
         / \
        3   10     <-- 10 has only a right child
       / \   \
      1   6   14   <-- 14 is the right child of 10

         / \ 
        4   7      <-- 4 and 7 are children of 6

*(In-order traversal of this BST yields: `1, 3, 4, 6, 7, 8, 10, 14`)*

### Properties

* Enables efficient search, insertion, and deletion of nodes based on their key value (on average).
* **In-order traversal always yields the nodes in sorted (ascending) order.**

### Operations

* **Search(value):**
    1.  Start at the root.
    2.  Compare `value` with the current node's key.
    3.  If equal, found.
    4.  If `value` < current key, go left.
    5.  If `value` > current key, go right.
    6.  Repeat until found or a null pointer is reached (not found).

* **Insert(value):**
    1.  Search for the position where `value` *should* be, as if performing a search.
    2.  When a null pointer is reached, insert the new node with `value` there.

* **Delete(value):**
    1.  Find the node to delete.
    2.  **Case 1: Node is a leaf.** Remove it (set parent's child pointer to null).
    3.  **Case 2: Node has one child.** Replace the node with its child (update parent's child pointer).
    4.  **Case 3: Node has two children.**
        * Find the node's **in-order successor** (the smallest value in the right subtree) OR **in-order predecessor** (the largest value in the left subtree).
        * Copy the successor's (or predecessor's) value into the node being deleted.
        * Delete the successor (or predecessor) node (this deletion will fall under Case 1 or Case 2).

### Complexity for BST Operations (Search, Insert, Delete)

Complexity depends directly on the **height ($h$)** of the tree.

* **General Time Complexity:** $O(h)$

* **Average Case (Reasonably Balanced Tree):**
    * Nodes are inserted in a relatively random order, leading to a balanced tree.
    * The height $h$ is proportional to $\log n$, so $h \approx O(\log n)$.
    * **Average Time Complexity:** $O(\log n)$

* **Worst Case (Skewed Tree):**
    * Nodes are inserted in sorted or reverse-sorted order, making the tree degenerate into a linked list.
    * The height $h$ becomes proportional to $n$, so $h \approx O(n)$.
    * **Worst-Case Time Complexity:** $O(n)$

* **Note on Balancing:** To avoid the $O(n)$ worst case, **self-balancing BSTs** (like AVL trees, Red-Black trees) perform rotations during insertion/deletion to guarantee a height of $O(\log n)$, ensuring $O(\log n)$ time complexity for all operations, albeit with higher constant factors due to the balancing overhead.

## BT vs. BST Summary

| Feature             | Binary Tree (BT)                      | Binary Search Tree (BST)                       |
| :------------------ | :------------------------------------ | :--------------------------------------------- |
| **Node Property** | Max 2 children                        | Max 2 children + Order (Left < Root < Right) |
| **Value Ordering** | None specific                         | Values are ordered                             |
| **Search `value`** | Requires full traversal ($O(n)$)      | Efficient on average ($O(\log n)$), $O(n)$ worst |
| **Insert `value`** | Location specific ($O(n)$ to find)    | Efficient on average ($O(\log n)$), $O(n)$ worst |
| **Delete `value`** | Requires search ($O(n)$)             | Efficient on average ($O(\log n)$), $O(n)$ worst |
| **In-Order Result** | Application specific                  | **Sorted** sequence                            |
| **Primary Use** | Representing hierarchy, Expressions   | Searching, Sorting, Dynamic Sets             |

## Applications

* **Binary Trees (General):**
    * Expression Trees (evaluating arithmetic expressions)
    * Huffman Coding Trees (data compression)
    * Abstract Syntax Trees (in compilers and interpreters)
    * Hierarchical data representation (e.g., file systems - simplified view)
* **Binary Search Trees:**
    * Implementing dynamic sets, maps, and dictionaries.
    * Symbol tables in programming language implementations.
    * Database indexing (though more complex B-trees/B+ trees are often used for disk-based storage).


```python
import time

def measure_time(sort_function, arr):
    start_time = time.time()
    sort_function(arr)
    return time.time() - start_time
    
# Binary Tree
class TreeNode:
    def __init__(self, val, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)


#      1
#     / \
#    2   3
#   / \   \
#  4  5    10

A = TreeNode(1) # Root
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)

A.left = B
A.right = C
B.left = D
B.right = E
C.right = F

print(A)
```

```python
# Recursive Pre Order (DFS) - Time: O(n), Space: O(n)
def pre_order(node):
    if not node:
        return
    print(node)
    pre_order(node.left)
    pre_order(node.right)

#pre_order(A)
print(f"Function: {measure_time(pre_order,A):.5f}")

```

```python
# Recursive In Order (DFS) - Time: O(n), Space: O(n)
def in_order(node):
    if not node:
        return
    in_order(node.left)
    print(node)
    in_order(node.right)

print(f"Function: {measure_time(in_order,A):.5f}")
```

```python
# Recursive post Order (DFS) - Time: O(n), Space: O(n)
def post_order(node):
    if not node:
        return
    post_order(node.left)
    post_order(node.right)
    print(node)

print(f"Function: {measure_time(post_order,A):.5f}")
```

```python
# Using a iterative pre order (DFS) - Time: O(n), Space: O(n)
def pre_order_itr(node):
    s = [node]
    
    while s:
        node = s.pop()
        print(node)
        if node.right:
            s.append(node.right)
        if node.left:
            s.append(node.left)

print(f"Function: {measure_time(pre_order_itr,A):.5f}")
```

```python
# BFS - level order traversal - Time: O(n), space: O(n)

from collections import deque

def level_order(node):
    q = deque()
    q.append(node)

    while q:
        node = q.popleft()
        print(node)
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)

print(f"Function: {measure_time(level_order,A):.5f}")
```

```python
# Check if value exist using DFS - Time: O(n), space: O(n)
def search(node, target):
    if not node:
        return False
    
    if node.val == target:
        return True
    
    return search(node.left, target) or search(node.right, target)

search(A, 5)
```

```python
# BST

#        5
#      /   \
#     1     8
#    / \   / \
#  -1  3  7  10

A2 = TreeNode(5)
B2 = TreeNode(1)
C2 = TreeNode(8)
D2 = TreeNode(-1)
E2 = TreeNode(3)
F2 = TreeNode(7)
G2 = TreeNode(10)

A2.left, A2.right = B2, C2
B2.left, B2.right = D2, E2
C2.left, C2.right = F2, G2

print(f"Function: {measure_time(level_order,A2):.5f}")

```

```python
# BST Search - Time: O(log n), Space: O(log n)

def search_bst(node, target):
    if not node:
        return False
    if node.val == target:
        return True
    
    if target < node.val:
        return search_bst(node.left, target)
    else:
        return search_bst(node.right, target)

search_bst(A2,3)
```
