## Part 2: Heaps & Priority Queues

### 5. Introduction: Relationship

**Priority Queue (PQ):** ADT defining `insert`, `peek`, `extract` by priority.  
**Heap:** Data structure (typically a Binary Heap) that efficiently implements a PQ.

---

### 6. Priority Queue (ADT)

#### 6.1 Definition

A collection where each element has an associated priority. Removal is based on highest (or lowest) priority.

#### 6.2 Conceptual Operations

- `insert(element, priority)`  
- `peek()` (find_max / find_min)  
- `extract()` (extract_max / extract_min)  
- *(Optional)* `change_priority(element, new_priority)`

#### 6.3 Analogy

Hospital triage: treat patients by severity, not arrival order.

---

### 7. Heap (Binary Heap)

#### 7.1 Definition

A complete binary tree stored in an array satisfying the heap property:

- **Max‑Heap:** Parent ≥ children (root is maximum).  
- **Min‑Heap:** Parent ≤ children (root is minimum).

#### 7.2 Array Representation

For node at index `i` (0‑based):

- **Parent:** `⌊(i − 1) / 2⌋`  
- **Left Child:** `2i + 1`  
- **Right Child:** `2i + 2`

#### 7.3 Examples

**Max‑Heap**  

        10
       /  \
      8    9
     / \  / \
    5  4  3  7

Array: [10, 8, 9, 5, 4, 3, 7]

**Min_Heap**  

         1
       /  \
      3    2
     / \  / \
    5   4 8  6

Array: [1, 3, 2, 5, 4, 8, 6]

### 8. Heap Operations (Max‑Heap)

- **Peek**  
  - Return `array[0]`  
  - **Time Complexity:** O(1)

- **Insert**  
  1. Append element to end of array.  
  2. **Heapify‑Up:** Swap with parent while > parent.  
  - **Time Complexity:** O(log n)

- **Extract‑Max**  
  1. Save root element.  
  2. Move last element to root, shrink heap.  
  3. **Heapify‑Down:** Swap with larger child while < child.  
  - **Time Complexity:** O(log n)

- **Build‑Heap**  
  - Call **Heapify‑Down** from the last non‑leaf node up to the root.  
  - **Time Complexity:** O(n)

---

### 9. Complexity Comparison for Priority Queues

| Operation         | Binary Heap | Sorted Array     | Unsorted Array |
|-------------------|-------------|------------------|----------------|
| Insert            | O(log n)    | O(n)             | O(1)           |
| Find Min/Max      | O(1)        | O(1)             | O(n)           |
| Extract Min/Max   | O(log n)    | O(1)* / O(n)     | O(n)           |
| Build             | O(n)        | O(n log n)       | O(1)           |

\* Extract from a sorted array is O(1) if removing from the end; O(n) if removing from the front.

---

### 10. Applications (Heaps & PQs)

- **OS Scheduling:** Task prioritization  
- **Graph Algorithms:** Dijkstra’s shortest path, Prim’s MST  
- **Heap Sort:** O(n log n) sorting algorithm  
- **Event Simulations:** Managing next event by time  
- **Data Compression:** Huffman coding  
- **Dynamic Scenarios:** Any scenario requiring repeated min/max access  


```python
# Build min heap (heapify)
# Time: O(n), space: O(1)

A = [-4, 3, 1, 0, 2 ,5, 10, 8, 12 ,9]

import heapq
heapq.heapify(A)

A
```

```python
# Heap push (inser element) - O(log n)

heapq.heappush(A,4)
A
```

```python
# Heap pop (Extract min) - O(log n)

min = heapq.heappop(A)

A, min
```

```python
# Heap Sort - O(n log n), space: O(n)
# NOTE: O(1) space is possible via swapping but complex

def heapSort(arr):
    heapq.heapify(arr)
    n = len(arr)
    new_list = [0] * n

    for i in range(n):
        min = heapq.heappop(arr)
        new_list[i] = min

    return new_list

heapSort([1, 3, 5, 7, 9, 2, 4, 6, 8, 0])
#heapSort(A)
```

```python
# Heap push pop - Time: O(log n)
heapq.heappushpop(A, 99)
A
```

```python
# Peek at min: Time O(1)
A[0]
```

```python
# Max heap
B = [-4, 3, 1, 0, 2 ,5, 10, 8, 12 ,9]
n = len(B)

for i in range(n):
    B[i] = -B[i]

heapq.heapify(B)
B
```

```python
larget = -heapq.heappop(B)
larget
```

```python
heapq.heappush(B, -7) #Insert 7 into max heap
B
```

```python
# Build heap from scratch

C = [-5, 4, 2, 1, 7, 0, 3]
heap = []

for x in C:
    heapq.heappush(heap, x)
    print(heap, len(heap))
```

```python
# Putting tuples of items on the heap
D = [5, 4, 3, 5, 4, 3, 5, 5, 4]

from collections import Counter

counter = Counter(D)

counter
```

```python
heap = []
for k, v in counter.items():
    print(k, v)
    heapq.heappush(heap, (v, k))

heap
```
