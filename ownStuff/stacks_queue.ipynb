### Stacks & Queue



```python
# Stack - LIFO
s = []
print(s)

#Append to top of stack - O(1)
s.append(5)
s.append(4)
s.append(3)
print(s)

#Pop from stack - O(1)
x = s.pop()
print(x)
print(s)

# Peek the top of stack - O(1)
print(s[-1])

# Ask if somnething is in stack - O(1)
if s:
    print(True)
```

```python
# Queue - FIFO

from collections import deque

q = deque()

print(q)

# To enqueue - Add element to the right - O(1)
q.append(5)
q.append(6)
q.append(7)
print(q)

# To Dequeue - Pop left - Remove element from the left - O(1)
q.popleft()
print(q)

# Peek from left side - O(1)
q[0]

# Peek from the right side - O(1)
q[-1]
```
