### Binary Search

- 2 Ways
    - Traditional ones using M = L + [(R-L) // 2]
    - Over under
        
- Time Complexity - O(log n)
- Space complexity - O()


```python
A = [-3, -1, 0, 1, 4, 7]

# Using approach 1
# Traditional Binary Search - Looking up if number is in array:
# Time: O(log n)
# Space: O(1)

def binary_search(arr, target):
    N = len(A)
    L = 0
    R = N - 1

    while L <= R:
        M = L + ((R-L) // 2)
        if(arr[M] == target):
            return True
        elif(arr[M] > target):
            R = M - 1
        else:
            L = M + 1

    return False

binary_search(A,4)
```

```python
# Using approach 2
# Based on a condition
# [F, F, F, F, T, T, T, T]
B = [False, False, False, False, True, True, True]

def binary_search_condition(arr):
    N = len(arr)
    L = 0
    R = N - 1

    while L < R:
        M = L + ((R-L) // 2)
        
        if(arr[M]):
            R = M
        else:
            L = M + 1
    return L

binary_search_condition(B)
```
